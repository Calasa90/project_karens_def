```python
#KARENS_API_KEY

# Importamos las herramientas necesarias para cargar y leer las variables del archivo .env
from dotenv import load_dotenv
import os

# Cargamos las variables guardadas en el archivo .env
load_dotenv()

# Guardamos la API key en una variable para poder utilizarla más adelante
api_key = os.getenv("KARENS_API_KEY")
```

```python
# Comprobamos que la API key se ha cargado correctamente sin mostrar su contenido
if api_key:
    print("API key cargada correctamente")
else:
    print("No se ha podido cargar la API key")
```

```python
# Importamos requests para poder realizar peticiones HTTP a la API
import requests

# Definimos el endpoint de The Cat API que utilizaremos para consultar las razas
url = "https://api.thecatapi.com/v1/breeds"
```

```python
# Creamos los headers para enviar nuestra API key de forma segura en la petición
headers = {
    "x-api-key": api_key
}

# Realizamos una petición GET al endpoint de razas
response = requests.get(url, headers=headers)

# Comprobamos el código de estado de la respuesta
print(response.status_code)
```

```python
# Convertimos la respuesta de la API de formato JSON a un objeto de Python
data = response.json()

# Comprobamos qué tipo de objeto hemos recibido
print(type(data))

# Comprobamos cuántos registros hemos obtenido
print(len(data))
```

```python
# Mostramos el primer elemento de la lista para analizar la estructura de los datos
data[0]
```

```python
# Mostramos las claves del primer registro para identificar todas las variables disponibles
data[0].keys()
```

```python
# Comprobamos si todos los registros tienen las mismas claves que el primer gato
same_keys = all(cat.keys() == data[0].keys() for cat in data)

print(same_keys)
```

```python
# Obtenemos las diferentes combinaciones de claves que aparecen en los registros
key_sets = {tuple(cat.keys()) for cat in data}

# Comprobamos cuántas estructuras diferentes existen
print("Número de estructuras diferentes:", len(key_sets))
```

```python
# Obtenemos el conjunto de todas las claves que aparecen en la API
all_keys = set().union(*(cat.keys() for cat in data))

# Mostramos todas las variables encontradas
print("Todas las claves:")
print(all_keys)

# Comprobamos qué claves faltan en cada registro
for cat in data:
    missing_keys = all_keys - set(cat.keys())

    if missing_keys:
        print(cat["name"], "-> faltan:", missing_keys)


#Eso significa que las variables que de verdad nos interesan para el análisis,
# como name, life_span, temperament, origin, weight, height y breed_group, sí aparecen en todos los registros.
```

```python
# Contamos en cuántos registros aparece cada una de las claves
key_counts = {
    key: sum(key in cat for cat in data)
    for key in all_keys
}

# Mostramos el número de registros que contienen cada variable
for key, count in key_counts.items():
    print(f"{key}: {count}/{len(data)}")
```

```python
# Creamos una lista vacía donde iremos guardando todos los animales
all_pets = []

# Empezamos sin cursor para obtener la primera página
cursor = None

while True:
    
    # Si tenemos cursor, lo enviamos como parámetro
    if cursor:
        params = {"cursor": cursor}
    else:
        params = {}

    # Realizamos la petición a la API
    response = requests.get(url_2, params=params)

    # Convertimos la respuesta a JSON
    page_data = response.json()

    # Añadimos los animales de esta página a nuestra lista
    all_pets.extend(page_data["pets"])

    # Mostramos el progreso
    print(
        f'Página {page_data["page"]}/{page_data["pages"]} - '
        f'Animales acumulados: {len(all_pets)}'
    )

    # Guardamos el cursor de la siguiente página
    # .get() devuelve None si "next_cursor" no existe
    cursor = page_data.get("next_cursor")

    # Si no existe siguiente cursor, hemos llegado al final
    if not cursor:
        break
```

```python
# Contamos cuántos gatos hay entre todos los animales obtenidos
total_cats = sum(
    pet["species"] == "Cat"
    for pet in all_pets
)

print("Total de gatos:", total_cats)
```

```python
# Comprobamos la información de paginación de la API de Zaragoza
print("Total de animales:", data_3["totalCount"])
print("Registro inicial:", data_3["start"])
print("Registros por petición:", data_3["rows"])
print("Animales recibidos:", len(data_3["result"])) 
```

```python
# Creamos una lista vacía donde iremos guardando todos los animales
all_pets_zaragoza = []

# Empezamos desde el primer registro
start = 0

# Indicamos cuántos registros queremos obtener en cada petición
rows = 50

while start < data_3["totalCount"]:

    # Definimos los parámetros de la petición
    params = {
        "start": start,
        "rows": rows
    }

    # Realizamos la petición a la API
    response = requests.get(url_3, params=params)

    # Convertimos la respuesta a JSON
    page_data = response.json()

    # Añadimos los animales obtenidos a nuestra lista
    all_pets_zaragoza.extend(page_data["result"])

    # Mostramos el progreso
    print(
        f'Registro {start} - '
        f'Animales acumulados: {len(all_pets_zaragoza)}'
    )

    # Avanzamos 50 registros para la siguiente petición
    start += rows
```

```python
# Comprobamos el número total de animales descargados
print("Total de animales descargados:", len(all_pets_zaragoza))
```

```python
# Contamos cuántos animales de todos los registros son de especie felina
total_cats_zaragoza = sum(
    animal["especie"] == "Felina"
    for animal in all_pets_zaragoza
)

print("Total de gatos:", total_cats_zaragoza)
```
